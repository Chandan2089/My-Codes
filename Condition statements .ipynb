```python
# 26. Strong Number

n = int(input("Enter the Number : "))

temp = n
s = 0

while n > 0 :
    d = n % 10
    n = n // 10

    fact = 1
    for i in range(1, d + 1):
        fact = fact * i

    s = s + fact

if s == temp:
    print("It's a Strong Number!")
else:
    print("It's not a Strong Number!")
```

```python
# 32. Kaprekar Routine

n = int(input("Enter a 4 digit number : "))         # Need to see it again 
m = 0
while n != 6174:

    # Extract digits
    a = n // 1000
    b = (n // 100) % 10
    c = (n // 10) % 10
    d = n % 10

    # Arrange digits in ascending order
    if a > b:
        a, b = b, a
    if a > c:
        a, c = c, a
    if a > d:
        a, d = d, a
    if b > c:
        b, c = c, b
    if b > d:
        b, d = d, b
    if c > d:
        c, d = d, c

    small = a * 1000 + b * 100 + c * 10 + d
    large = d * 1000 + c * 100 + b * 10 + a

    n = large - small
    m = m + 1

    print(large, "-", small, "=", n)
print("Total no. of Operations took place are : " , m)
```

```python
# 34. Binary to Decimal with validation

n = int(input("Enter the binary number : "))

temp = n
decimal = 0
power = 0
valid = 1

# Validation
while temp > 0:
    d = temp % 10
    temp = temp // 10

    if d != 0 and d != 1:
        valid = 0
        break

# Conversion
if valid == 1:
    temp = n

    while temp > 0:
        d = temp % 10
        temp = temp // 10

        decimal = decimal + d * (2 ** power)
        power = power + 1

    print("Decimal:", decimal)

else:
    print("Invalid binary number")
```

```python
# 37. Longest consecutive character run 

s = "aaabbccccdaa"

max_char = s[0]
max_run = 1
current_run = 1

for i in range(1, len(s)):
    if s[i] == s[i - 1]:
        current_run = current_run + 1
    
    else:
        current_run = 1
        
    if current_run > max_run:
        max_run = current_run
        max_char = s[i]

print("Character:", max_char)
print("Longest run:", max_run)
```

```python
# 38. Run-length Encoding

s = "aaabbccccdaa"

count = 1
result = ""

for i in range(1, len(s)):
    if s[i] == s[i - 1]:
        count = count + 1
    else:
        result = result + s[i - 1] + str(count)
        count = 1

# Add the last character and its count
result += s[-1] + str(count)

print(result)
```

```python
# 41. GCD and LCM

# Read two numbers
a = int(input("Enter 1st No. : "))
b = int(input("Enter 2nd No. : "))

# Find GCD
gcd = 1

for i in range(1, min(a, b) + 1):
    if a % i == 0 and b % i == 0:
        gcd = i

# Find LCM
lcm = (a * b) // gcd

# Print results
print("GCD:", gcd)
print("LCM:", lcm)
```

```python
# 43. Rectangle 

rows = int(input("Enter rows: "))
cols = int(input("Enter columns: "))

for i in range(rows):
    for j in range(cols):
        if i == 0 or i == rows - 1 or j == 0 or j == cols - 1:
            print("*", end="")
        else:
            print(" ", end="")
    print()
```

```python
# 45. Number Reduction Game

n = int(input("Enter the Input : "))

s = 0

while n > 0:
    print(n, end=" ")

    if n % 2 == 0:
        n = n // 2
    else:
        n = n - 1

    s = s + 1

print(0)
print("Steps: ", s)
```
